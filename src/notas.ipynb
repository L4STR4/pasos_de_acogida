"He utilizado el dataset de Austin por su volumen (+156.000 registros) para entender la analítica de rotación en refugios, pero en España la realidad descrita por el estudio de la Fundación Affinity presenta diferencias críticas:

1. Rotación y Tiempo de Estancia: En Austin la mediana de estancia es de 5,8 días gracias a su tasa de rotación y política No-Kill con red de traslados. En España, un perro adulto pasa una media de 8 a 12 meses en una protectora antes de ser adoptado.
2. Identificación por Microchip: En Austin la tasa de devolución al dueño por chip es del 15,6%. En España, menos del 27% de los perros y solo el 4% de los gatos que entran a un refugio están identificados, lo que convierte una pérdida temporal en un abandono permanente.
3. El Factor Estacional: En ambos contextos se confirma que la primavera/verano es el momento crítico, no solo por vacaciones, sino por las camadas no deseadas fruto de la falta de esterilización."
